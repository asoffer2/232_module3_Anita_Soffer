## Leetcode 217 :
Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.

````python
# answer
def int_array(nums):
  num_set = set(nums)
  if len(num_set) == len(nums):
    return False
  else:
    return True

print(int_array([1,2,2,3,4,5]))
print(int_array([1,2,3,4]))
````

Starting with any positive integer, replace the number by the sum of the squares of its digits.
Repeat the process until the number equals 1 (where it will stay), or it loops endlessly in a cycle which does not include 1.
Those numbers for which this process ends in 1 are happy.

Explanation:

Example 2:

````python
# solution
'''def happy_number(num):
  num_set = set()
  all_nums = []
  final = 0
  while final != 1:
    if final == 1:
      return True
    num_list = []
    str_num = str(num)
    for num in str_num:
      num_list.append(int(num))
    for num in num_list:
      squared = num**2
      final += squared
    if final == 1:
      return True
    else:
      num_set.update(final)
      all_nums.append(final)
      if len(num_set) == len(all_nums):
        return False

print(happy_number(19))'''

# answer
def cycle(num):
  if not isinstance(num, int):
    return None

  seen = set()

  while num != 1 and num not in seen: # exit if the number is one or not seen
    seen.add(num)
    num = sum(int(d)**2 for d in str(num))
    print(num, end = ', ')

  return num == 1

print(cycle(10))



````
